## 1. Qué es un secreto y por qué importa.

* Una **clave de API** es una contraseña que identifica tu cuenta ante un servicio (*Gemini*, *LangSmith*). Quien la tenga puede consumir tu cuota y generar cargos a tu nombre.
* Un **secreto de *Codespaces*** es una variable cifrada que guardas en tu cuenta de GitHub. Cuando abres un *Codespace* de un repositorio autorizado, GitHub la inyecta como **variable de entorno** (por ejemplo, `GOOGLE_API_KEY`). El código la lee con `os.environ["GOOGLE_API_KEY"]` y la clave nunca se escribe en un archivo.
* Este repositorio es **público**. Una clave escrita en un *notebook*, aunque la borres después, queda en el historial de Git y cualquiera puede encontrarla.

**Regla del curso:** las claves solo viven en los secretos de *Codespaces* (o en un archivo `.env` local que Git ignora). Nunca en un *notebook*, nunca en un mensaje, nunca en una captura de pantalla.

## 4. Guardar las claves como secretos de *Codespaces*.

Hazlo una vez por clave. Los nombres deben escribirse **exactamente** como aparecen en la tabla de la sección 2.

1. En GitHub, haz clic en tu foto de perfil (esquina superior derecha) → **Settings**.
2. En la barra lateral, en la sección **Code, planning, and automation**, haz clic en **Codespaces**.
3. A la derecha de **Codespaces secrets**, haz clic en **New secret**.
4. En **Name**, escribe `GOOGLE_API_KEY`.
5. En **Value**, pega la clave de *Gemini*. Sin comillas ni espacios al inicio o al final.
6. En **Repository access**, selecciona el repositorio del curso: tu copia (*fork*) de `py501` o `PythonistaMX/py501`, según desde dónde vayas a crear el *Codespace*.
7. Haz clic en **Add secret**.
8. Repite los pasos 3 a 7 con `LANGSMITH_API_KEY` y, si las tienes, con `OPENAI_API_KEY` y `ANTHROPIC_API_KEY`.

**Reglas de los nombres** (las aplica GitHub): solo letras, números y guion bajo; no pueden empezar con número ni con `GITHUB_`; no distinguen mayúsculas de minúsculas. Puedes guardar hasta 100 secretos de 48 KB cada uno.

### Alternativa: capturarlas al crear el *Codespace*.

El archivo `.devcontainer/devcontainer.json` declara estos secretos como **recomendados**. Si creas el *Codespace* con **Code → Codespaces → `…` → New with options**, GitHub muestra los campos para capturarlos ahí mismo. Lo que escribas se guarda en tus secretos personales y no tendrás que volver a capturarlo.

## 6. Verificar sin mostrar las claves.

> ⚠️ **Nunca** ejecutes `print(os.environ["GOOGLE_API_KEY"])` ni `!echo $GOOGLE_API_KEY`: la clave quedaría en la salida de la celda, y la salida se guarda **dentro del *notebook***. Las celdas de esta sección solo informan si la clave existe y si funciona.

* La siguiente celda indicará qué claves existen como variables de entorno, sin mostrar su valor.

```python
import os

CLAVES = {
    "GOOGLE_API_KEY": "obligatoria",
    "LANGSMITH_API_KEY": "recomendada",
    "OPENAI_API_KEY": "opcional",
    "ANTHROPIC_API_KEY": "opcional",
}
{nombre: f"{'✅ configurada' if os.environ.get(nombre) else '❌ falta'} ({tipo})" for nombre, tipo in CLAVES.items()}
```

**Interpretación:** `GOOGLE_API_KEY` debe aparecer como ✅. Si aparece ❌, revisa la sección 7 antes de continuar: las siguientes celdas lo informarán, pero no podrán probar la clave.

* La siguiente celda cargará `.env` si existe. En *Codespaces* no debe existir: ahí se usan los secretos.

```python
from dotenv import load_dotenv

RAIZ = next(p for p in (os.curdir, os.pardir) if os.path.isdir(os.path.join(p, "data", "politicas")))
cargado = load_dotenv(os.path.join(RAIZ, ".env"))
"Se cargó .env (ejecución local)" if cargado else "No hay .env: se usan los secretos de Codespaces"
```

* La siguiente celda hará una llamada mínima a *Gemini* y traducirá los errores más comunes a una instrucción concreta.

```python
def probar_gemini() -> str:
    if not (os.environ.get("GOOGLE_API_KEY") or os.environ.get("GEMINI_API_KEY")):
        return "❌ No hay GOOGLE_API_KEY: crea el secreto (sección 4) y reabre el Codespace (sección 5)."
    from langchain.chat_models import init_chat_model
    try:
        texto = init_chat_model("google_genai:gemini-3.5-flash").invoke("Responde solo con la palabra: listo").text
        return f"✅ Gemini respondió: {texto!r}"
    except Exception as exc:  # se clasifica el error sin mostrar la clave
        mensaje = str(exc)
        if any(s in mensaje for s in ("API key not valid", "API_KEY_INVALID", "PERMISSION_DENIED", "401", "403")):
            return "❌ La clave no es válida o no está restringida a la Gemini API: revisa la sección 3.1 y la sección 7."
        if any(s in mensaje for s in ("429", "RESOURCE_EXHAUSTED", "quota")):
            return "⚠️ La clave funciona, pero se agotó el límite del nivel gratuito: espera unos minutos."
        return f"❌ Error inesperado ({type(exc).__name__}): {mensaje[:200]}"


probar_gemini()
```

* La siguiente celda comprobará que *LangSmith* acepta la clave.

```python
def probar_langsmith() -> str:
    if not os.environ.get("LANGSMITH_API_KEY"):
        return "⚠️ No hay LANGSMITH_API_KEY: los notebooks funcionan, pero sin trazas ni experimentos (sección 3.2)."
    from langsmith import Client
    try:
        next(iter(Client().list_projects(limit=1)), None)
        return f"✅ LangSmith responde. Las trazas irán al proyecto {os.environ.get('LANGSMITH_PROJECT', 'default')!r}."
    except Exception as exc:
        return f"❌ LangSmith rechazó la clave o no responde ({type(exc).__name__}): revisa la sección 3.2."


probar_langsmith()
```

* La siguiente celda revisará que ningún archivo del repositorio contenga algo con forma de clave de API (una auditoría rápida que conviene repetir antes de cada `git push`).

```python
import json
import re

PATRONES = {"Google": r"AIza[0-9A-Za-z_\-]{20,}", "LangSmith": r"lsv2_[0-9A-Za-z_]{20,}",
            "OpenAI": r"sk-[0-9A-Za-z_\-]{20,}", "Anthropic": r"sk-ant-[0-9A-Za-z_\-]{20,}"}
hallazgos = []
for carpeta, _, archivos in os.walk(RAIZ):
    if ".git" in carpeta:
        continue
    for archivo in archivos:
        if archivo.endswith((".ipynb", ".py", ".md", ".json", ".txt", ".yml")):
            ruta = os.path.join(carpeta, archivo)
            with open(ruta, encoding="utf-8", errors="ignore") as f:
                contenido = f.read()
            hallazgos += [(ruta, proveedor) for proveedor, patron in PATRONES.items() if re.search(patron, contenido)]
hallazgos or "✅ No se encontraron claves escritas en los archivos del repositorio."
```

* La siguiente celda confirmará que Git ignora el archivo `.env`.

```python
import subprocess

resultado = subprocess.run(["git", "check-ignore", "-q", os.path.join(RAIZ, ".env")])
"✅ .env está en .gitignore: Git nunca lo subirá." if resultado.returncode == 0 else "❌ .env NO está ignorado: no lo crees hasta corregir .gitignore."
```

**💡 Ideas clave de la sección 6:**
* Se verifica que la clave **existe** y que **funciona**, sin imprimir su valor.
* Antes de subir cambios, se revisa que ningún archivo contenga una clave.
* `.env` solo existe en ejecución local y Git lo ignora.

## 7. Errores comunes.

| Síntoma | Causa probable | Solución |
|---|---|---|
| `RuntimeError: Falta GOOGLE_API_KEY` o ❌ en la sección 6 | El secreto no existe, tiene otro nombre o no tiene acceso al repositorio | Revisa el nombre exacto y la sección **Repository access** del secreto |
| El secreto existe pero el *Codespace* no lo ve | El *Codespace* se inició antes de crear el secreto | Detén y vuelve a abrir el *Codespace* (sección 5) |
| `API key not valid` o `PERMISSION_DENIED` | Clave mal copiada, revocada o no restringida a la *Gemini API* | Crea una clave nueva en *AI Studio* y actualiza el secreto (**Update** junto al secreto) |
| `429` o `RESOURCE_EXHAUSTED` | Se agotó el límite del nivel gratuito | Espera unos minutos, ejecuta menos celdas o pide la clave de respaldo al instructor |
| Las celdas de *LangSmith* fallan | Falta `LANGSMITH_API_KEY` o es inválida | Agrega o actualiza el secreto (secciones 3.2 y 4) |
| Creaste el *Codespace* desde tu *fork* y no ve los secretos | Diste acceso a `PythonistaMX/py501`, no a tu *fork* | Agrega tu *fork* en **Repository access** |

## 8. Si una clave se filtra.

1. **Revócala de inmediato:** en *AI Studio* (o en *LangSmith*), elimina la clave.
2. Crea una clave nueva.
3. Actualiza el secreto en **Settings → Codespaces → Codespaces secrets → Update**.
4. Revisa el uso y la facturación del proyecto para detectar consumo que no reconozcas.

Borrar la clave del *notebook* **no basta**: sigue en el historial de Git y en cualquier copia que alguien haya hecho.
